```python
# Stape 1 Import Libraries
import pandas as pd
import numpy as np

# Step 2 Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Display Settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)

# Ignore Warnings
import warnings
warnings.filterwarnings('ignore')
```

```python
# Step 2 Load Dataset
df=pd.read_csv("Electricity Consumption.csv")
```

```python
# Step 3 Data Understanding

print("="*60)
print("DATA UNDERSTANDING")
print("="*60)

# Data Preview

print("\nFirst 5 Records")
display(df.head())

print("\nLast 5 Records")
display(df.tail())

# 2. Data Shape 

print("Rows  :", df.shape[0])
print("Columns:", df.shape[1])

#3. Column Name
print("\nColumn Names")
print(df.columns.tolist)

#4 Data set information
print("\nDataset Information")
df.info()

#5 Statistical Summary
print("\nStatistical Summary")
display(df.describe())

# 6 Data Types
print("\nData Types")
display(df.dtypes)

# 7 Missing Values

print("\nMissing Values")
missing_values = df.isnull().sum()
display(missing_values)
print(
    "\nTotal Missing Values:",
    missing_values.sum()
)

#8 Duplicate Records

duplicates = df.duplicated().sum()
print(
    "\nTotal Duplicate Records:",
    duplicates
)

# 9 Unique values
print("\nUnique Values in Each Column")
for col in df.columns:
    print(f"\nColumns :  {col}")
    print(df[col].unique())
    print("-"*50)

# 10. Count Unique Values
print("\nNumber of Unique Values")

display(df.nunique())


# 11. Numerical Columns
numerical_cols = df.select_dtypes(
    include=['int64', 'float64']
).columns

print("\nNumerical Columns")

print(numerical_cols.tolist())


# 12. Categorical Columns

categorical_cols = df.select_dtypes(
    include=['object']
).columns

print("\nCategorical Columns")

print(categorical_cols.tolist())


# 13. Frequency Distribution

print("\nFrequency Distribution")

for col in categorical_cols:

    print(f"\nColumn : {col}")

    print(df[col].value_counts())

    print("-"*50)


# 14. Furnishing Status Distribution
plt.figure(figsize=(8,5))

sns.countplot(
    x='Unit Consumption',
    data=df,
    palette='Set2'
)

plt.title("Unit Consumption Status Distribution")

plt.xlabel("Unit Consumption Status")

plt.ylabel("Count")

plt.show()


# 15. Correlation Matrix

print("\nCorrelation Matrix")

corr_matrix = df.corr(
    numeric_only=True
)

display(corr_matrix)


# 16. Correlation Heatmap

plt.figure(figsize=(10,8))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title(
    "Correlation Heatmap"
)

plt.show()


# 17. Summary Report
print("\nSUMMARY REPORT")

print("-"*50)

print(
    "Total Rows      :",
    df.shape[0]
)

print(
    "Total Columns   :",
    df.shape[1]
)

print(
    "Missing Values  :",
    df.isnull().sum().sum()
)

print(
    "Duplicate Rows  :",
    df.duplicated().sum()
)

print(
    "Numerical Cols  :",
    len(numerical_cols)
)

print(
    "Categorical Cols:",
    len(categorical_cols)
)

print("\nData Understanding Completed Successfully")
```

```python
# =====================================================
# STEP 4 : DATA CLEANING
# =====================================================

print("="*60)
print("DATA CLEANING")
print("="*60)

# -----------------------------------------------------
# Create Copy of Dataset
# -----------------------------------------------------

df_clean = df.copy()

print("\nOriginal Shape :", df.shape)

# -----------------------------------------------------
# 1. Check Missing Values
# -----------------------------------------------------

print("\nMissing Values Before Cleaning")

display(df_clean.isnull().sum())

print(
    "\nTotal Missing Values :",
    df_clean.isnull().sum().sum()
)

# -----------------------------------------------------
# 2. Handle Missing Values
# -----------------------------------------------------

# Numerical Columns

num_cols = df_clean.select_dtypes(
    include=['int64','float64']
).columns

for col in num_cols:

    if df_clean[col].isnull().sum() > 0:

        df_clean[col].fillna(
            df_clean[col].median(),
            inplace=True
        )

# Categorical Columns

cat_cols = df_clean.select_dtypes(
    include=['object']
).columns

for col in cat_cols:

    if df_clean[col].isnull().sum() > 0:

        df_clean[col].fillna(
            df_clean[col].mode()[0],
            inplace=True
        )

# -----------------------------------------------------
# 3. Verify Missing Values
# -----------------------------------------------------

print("\nMissing Values After Treatment")

display(df_clean.isnull().sum())

print(
    "\nTotal Missing Values :",
    df_clean.isnull().sum().sum()
)

# -----------------------------------------------------
# 4. Check Duplicate Records
# -----------------------------------------------------

duplicates = df_clean.duplicated().sum()

print(
    "\nDuplicate Records Before Removal :",
    duplicates
)

# -----------------------------------------------------
# 5. Remove Duplicate Records
# -----------------------------------------------------

df_clean.drop_duplicates(
    inplace=True
)

print(
    "\nShape After Duplicate Removal :",
    df_clean.shape
)

# -----------------------------------------------------
# 6. Standardize Text Data
# -----------------------------------------------------

cat_cols = df_clean.select_dtypes(
    include=['object']
).columns

for col in cat_cols:

    df_clean[col] = (
        df_clean[col]
        .astype(str)
        .str.lower()
        .str.strip()
    )

print(
    "\nText Standardization Completed"
)

# -----------------------------------------------------
# 7. Verify Unique Values
# -----------------------------------------------------

print("\nUpdated Categories")

for col in cat_cols:

    print(f"\n{col}")

    print(df_clean[col].unique())

# -----------------------------------------------------
# 8. Check Data Types
# -----------------------------------------------------

print("\nData Types")

display(df_clean.dtypes)

# -----------------------------------------------------
# 9. Outlier Check (IQR Method)
# -----------------------------------------------------

print("\nOutlier Summary")

for col in num_cols:

    Q1 = df_clean[col].quantile(0.25)

    Q3 = df_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR

    upper = Q3 + 1.5 * IQR

    outliers = df_clean[
        (df_clean[col] < lower) |
        (df_clean[col] > upper)
    ].shape[0]

    print(
        f"{col} : {outliers} outliers"
    )

# -----------------------------------------------------
# 10. Final Dataset Shape
# -----------------------------------------------------

print("\nFinal Dataset Shape")

print(df_clean.shape)

# -----------------------------------------------------
# 11. Dataset Preview
# -----------------------------------------------------

print("\nCleaned Dataset Preview")

display(df_clean.head())

# -----------------------------------------------------
# Cleaning Summary
# -----------------------------------------------------

print("\nDATA CLEANING SUMMARY")

print("-"*50)

print(
    "Rows :", df_clean.shape[0]
)

print(
    "Columns :",
    df_clean.shape[1]
)

print(
    "Missing Values :",
    df_clean.isnull().sum().sum()
)

print(
    "Duplicate Records :",
    df_clean.duplicated().sum()
)

print("\nData Cleaning Completed Successfully")
```

```python
# STEP 5 : ENCODE BINARY YES/NO COLUMNS


print("="*60)
print("ENCODE BINARY YES/NO COLUMNS")
print("="*60)


# Create Copy

df_encoded = df_clean.copy()


# Binary Columns
binary_cols = [
    'smart_meter',
    'Solar panel'
]

print("\nBinary Columns")

print(binary_cols)


# Check Original Values

print("\nOriginal Value Counts")

for col in binary_cols:

    print(f"\nColumn : {col}")

    print(df_encoded[col].value_counts())


# Encode Yes / No

for col in binary_cols:

    df_encoded[col] = df_encoded[col].map({
        'yes': 1,
        'no': 0
    })


# Verify Encoding

print("\nEncoded Values")

display(df_encoded[binary_cols].head())


# Check Data Types

print("\nEncoded Column Data Types")

print(df_encoded[binary_cols].dtypes)


# Check Unique Values
print("\nUnique Values After Encoding")

for col in binary_cols:

    print(
        f"{col} :",
        df_encoded[col].unique()
    )


# Dataset Preview
print("\nDataset Preview")

display(df_encoded.head())


# Shape Check

print("\nDataset Shape")

print(df_encoded.shape)


# Summary
print("\nEncoding Completed Successfully")

print(
    "Encoded Columns :",
    len(binary_cols)
)
```

```python
# STEP 6 : FEATURE ENGINEERING

print("="*60)
print("FEATURE ENGINEERING")
print("="*60)


# Create Copy

df_fe = df_encoded.copy()

print("\nOriginal Shape :", df_fe.shape)


# 1. Total smart_meter

df_fe['total smart_meter'] = (
    df_fe['Unit Consumption']
)

print("\nFeature Created : smart_meter")





# Check New Features


new_features = [

    'total smart_meter',

    'Unit Consumption'

]

print("\nNew Features")

print(new_features)


# Preview New Features

print("\nFeature Preview")

display(
    df_fe[new_features].head()
)


# Statistical Summary


print("\nFeature Statistics")

display(
    df_fe[new_features].describe()
)


# Select numeric columns only
numeric_df = df_fe.select_dtypes(include=['number'])

# Calculate correlation matrix
feature_corr = numeric_df.corr()

# Display it
print(feature_corr)

# Visualize Correlation
plt.figure(figsize=(8,5))

sns.heatmap(
    feature_corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title(
    "Feature Engineering Correlation Heatmap"
)

plt.show()


# Dataset Shape

print("\nDataset Shape After Feature Engineering")

print(df_fe.shape)


# Dataset Preview

print("\nDataset Preview")

display(df_fe.head())


# Summary


print("\nFEATURE ENGINEERING SUMMARY")

print("-"*50)

print(
    "Features Added :",
    len(new_features)
)

print(
    "Final Rows :",
    df_fe.shape[0]
)

print(
    "Final Columns :",
    df_fe.shape[1]
)

print("\nFeature Engineering Completed Successfully")
```

```python
# STEP 7 : ENCODE REMAINING CATEGORICAL VARIABLES

print("="*60)
print("ENCODE REMAINING CATEGORICAL VARIABLES")
print("="*60)

# Create Copy
df_model = df_fe.copy()

# Check Remaining Categorical Columns
print("\nCategorical Columns:")
categorical_cols = df_model.select_dtypes(include='object').columns
print(categorical_cols.tolist())

# Check available columns
print("\nAvailable Columns:")
print(df_model.columns.tolist())

# Create Month status column FIRST
if 'Month' in df_model.columns:

    df_model['Month status'] = df_model['Month'].apply(
        lambda x: 'High' if str(x) in ['Apr', 'May', 'Jun'] else 'Normal'
    )

    print("\nMonth Categories")
    print(df_model['Month status'].value_counts())

    # One-Hot Encoding
    df_model = pd.get_dummies(
        df_model,
        columns=['Month status'],
        drop_first=True
    )

else:
    print("\nERROR: 'Month' column not found in dataset.")

# Verify new columns
print("\nEncoded Columns Created")

new_cols = [
    col for col in df_model.columns
    if 'Month status' in col
]

print(new_cols)

# Dataset Information
print("\nDataset Information")
df_model.info()

# Remaining categorical columns
print("\nRemaining Object Columns")
remaining_cat = df_model.select_dtypes(include='object').columns
print(remaining_cat.tolist())

# Dataset Shape
print("\nDataset Shape")
print("Rows :", df_model.shape[0])
print("Columns :", df_model.shape[1])

# Preview
print("\nDataset Preview")
display(df_model.head())

# Save Dataset
df_model.to_csv(
    "Model_Ready_Dataset.csv",
    index=False
)

print("\nModel Ready Dataset Saved Successfully")
```

```python
#  8.FEATURE IMPORTANCE USING CORRELATION


corr_with_price = (
    df_model.corr(numeric_only=True)['Unit Consumption']
    .sort_values(ascending=False)
)

print("Correlation with Unit Consumption")

display(corr_with_price)

# Visualization

plt.figure(figsize=(10,6))

corr_with_price.drop('Unit Consumption').sort_values().plot(
    kind='barh'
)


plt.title("Feature Correlation with Unit Consumption")

plt.xlabel("Correlation")

plt.show()

```

```python
# STEP 9 : TRAIN-TEST SPLIT & FEATURE SCALING

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import pandas as pd
import joblib

print("="*60)
print("TRAIN-TEST SPLIT & FEATURE SCALING")
print("="*60)

# Separate Features and Target
X = df_model.drop('Unit Consumption', axis=1)
y = df_model['Unit Consumption']

print("\nOriginal Features Shape :", X.shape)
print("Target Shape            :", y.shape)

# Check Object Columns
print("\nObject Columns Found:")
object_cols = X.select_dtypes(include=['object']).columns.tolist()
print(object_cols)

# Remove Object Columns
X = X.select_dtypes(include=['number', 'bool'])

print("\nFeatures After Removing Text Columns")
print(X.shape)

# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTrain-Test Split Completed")

print("X_train Shape :", X_train.shape)
print("X_test Shape  :", X_test.shape)

print("y_train Shape :", y_train.shape)
print("y_test Shape  :", y_test.shape)

# Feature Scaling
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Convert Back To DataFrame
X_train_scaled = pd.DataFrame(
    X_train_scaled,
    columns=X.columns
)

X_test_scaled = pd.DataFrame(
    X_test_scaled,
    columns=X.columns
)

# Preview
print("\nScaled Training Dataset Preview")
display(X_train_scaled.head())

# Mean Check
print("\nFeature Means")
display(X_train_scaled.mean().round(2))

# Standard Deviation Check
print("\nFeature Standard Deviations")
display(X_train_scaled.std().round(2))

# Save Scaler
joblib.dump(
    scaler,
    "scaler.pkl"
)

print("\nScaler Saved Successfully")

# Save Processed Data
X_train_scaled.to_csv(
    "X_train_scaled.csv",
    index=False
)

X_test_scaled.to_csv(
    "X_test_scaled.csv",
    index=False
)

print("\nScaled Datasets Saved Successfully")

# Summary
print("\nTRAIN-TEST SPLIT SUMMARY")
print("-"*50)

print("Training Records :", len(X_train))
print("Testing Records  :", len(X_test))
print("Number of Features :", X.shape[1])

print("\nStep 9 Completed Successfully")
```

```python
# STEP 10 : DATA MODELING

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

print("="*60)
print("DATA MODELING")
print("="*60)


# 1. Linear Regression

lr_model = LinearRegression()

lr_model.fit(
    X_train_scaled,
    y_train
)

lr_pred = lr_model.predict(
    X_test_scaled
)

print("✓ Linear Regression Model Trained Successfully")


# 2. Decision Tree Regressor
dt_model = DecisionTreeRegressor(
    random_state=42
)

dt_model.fit(
    X_train_scaled,
    y_train
)

dt_pred = dt_model.predict(
    X_test_scaled
)

print("✓ Decision Tree Model Trained Successfully")


# 3. Random Forest Regressor
rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

rf_model.fit(
    X_train_scaled,
    y_train
)

rf_pred = rf_model.predict(
    X_test_scaled
)

print("✓ Random Forest Model Trained Successfully")


# Preview Predictions

comparison = pd.DataFrame({

    'Actual Price': y_test.values,

    'Linear Regression': lr_pred,

    'Decision Tree': dt_pred,

    'Random Forest': rf_pred

})

print("\nPrediction Comparison")

display(
    comparison.head()
)

print("\nAll Models Trained Successfully")
```
